```python
gastos = [
    ["Comida", 45.20], ["Transporte", 12.50], ["Ocio", 30.00], ["Comida", 67.80],
    ["Suministros", 85.00], ["Transporte", 8.75], ["Ocio", 55.20], ["Comida", 23.40],
    ["Suministros", 42.10], ["Salud", 60.00], ["Transporte", 15.00], ["Comida", 38.90],
    ["Ocio", 20.00], ["Salud", 25.50], ["Suministros", 33.30], ["Comida", 52.00],
    ["Transporte", 9.20], ["Ocio", 41.60], ["Salud", 18.75], ["Comida", 29.10],
]

gastos_dicc = {}
gasto_total = 0

for g in gastos:
    categoria = g[0]
    gasto = g[1]
    gastos_dicc[categoria] = gastos_dicc.get(categoria, 0) + gasto     # "coge lo que ya tenías acumulado para esta categoría (o 0 si es la primera vez),                                                                     
    gasto_total = gasto_total + gasto                                   # súmale el gasto nuevo, y guarda ese total actualizado en el diccionario, bajo esa misma categoría.

print(gastos_dicc)    
print(gasto_total)

categoria_mayor = None
gasto_max = 0

for categoria, gasto in gastos_dicc.items():
    if gasto > gasto_max:
        gasto_max = gasto
        categoria_mayor = categoria

print(f'La categoría con mayor gasto es -> {categoria_mayor}: {gasto_max:.2f} € ')
porcentaje = (gasto_max/gasto_total)*100
print(f'{porcentaje:.2f} % ')
    
```

Contiene al menos un número.

Un bucle que recorra la lista y muestre cada contraseña junto a si es válida o no.

Pista para la condición 2 (contiene al menos un número): puedes recorrer cada carácter de la contraseña con un for letra in contraseña: y comprobar con letra.isdigit() si ese carácter es un número.
```python
contrasenas = ["abc123", "password1", "hello", "MiClave2024", "12345", "seguridad9", "qwe", "contraseñaSegura8", "abcd", "test1234","pepepepepe"]

def es_valida(password):
    if len(password) < 8:
        return False
    tiene_numero = False
    for letra in password:
        if letra.isdigit():
            tiene_numero = True
    return tiene_numero

for password in contrasenas:
    resultado = es_valida(password)
    if not resultado:
        print(f'La contraseña {password} es inválida')
    else:
        print(f'La contraseña {password} es válida')
    
```

Escribe un programa que:

Encuentre cuál fue el turno (mañana/tarde/noche) que más cajas produjo en total, sumando esa columna a lo largo de los 7 días — sin usar max().

```python
produccion = [
    [120, 135, 98], [110, 140, 102], [125, 130, 95], [118, 145, 100],
    [130, 128, 110], [95, 100, 80], [90, 95, 75]
]

total_mañana = 0
total_tarde = 0
total_noche = 0 
dias_semana = ['Lunes', 'Martes', 'Miercoles', 'Jueves', 'Viernes', 'Sábado', 'Domingo']

for d, c in zip(dias_semana, produccion):
    mañana = c[0]
    tarde = c[1]
    noche = c[2]
    resultado_diario = mañana + tarde + noche
    print(f'Producción {d}: {resultado_diario} cajas')

    total_mañana = total_mañana + mañana
    total_tarde = total_tarde + tarde
    total_noche = total_noche + noche

nombre_turno_max = 'mañana'
turno_max = total_mañana

if total_tarde > turno_max:
    turno_max = total_tarde
    nombre_turno_max = 'tarde'
if total_noche > turno_max:
    turno_max = total_noche
    nombre_turno_max = 'noche'

print()
print(f'El turno con más producción fue la {nombre_turno_max} con {turno_max} cajas.')
print()
print(f'En los turnos de mañana se ha producido: {total_mañana}')
print(f'En los turnos de tarde se ha producido: {total_tarde}')
print(f'En los turnos de noche se ha producido: {total_noche}')
```

**Ejercicio 26: Comprensión de listas (concepto nuevo) - List comprehensions**

Usando comprensión de listas (no un for tradicional), crea una lista nueva llamada caros que contenga solo los precios mayores de 8€.

Python tiene una forma más compacta de escribir exactamente esto mismo, en una sola línea, llamada list comprehension:
- aprobados = [a for a in notas if a >= 5]

Se lee así: "quiero una lista con a, por cada a en notas, pero solo si a >= 5". Es la misma lógica del for+if+.append(), condensada.

```python
precios = [3.20, 8.50, 1.90, 12.30, 4.75, 6.60, 2.10, 15.00, 5.40, 9.90, 0.85, 7.25, 11.20, 3.95, 6.80, 2.50, 14.10, 4.20, 8.90, 1.50]

caros = [c for c in precios if c > 8] # la primera c actua como caros.append(c)
print(caros) 

# Lo anterior es lo mismo que:

for c in precios:
    if c > 8:
        caros.append(c)
        print(c)
```
