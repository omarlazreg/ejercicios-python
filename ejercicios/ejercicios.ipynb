**Ejercicio 10: Filtro de pedidos**

Una tienda online tiene registrado el importe (en euros) de los últimos pedidos del día. La empresa quiere aplicar un envío gratuito automático a los pedidos de 50€ o más.

Escribe:
Una función llamada envio_gratis que reciba un importe y devuelva True si es 50€ o más, False si no.
Un bucle que recorra pedidos y, usando esa función, cuente cuántos pedidos tienen envío gratis y cuántos no.
Al final, muestra ambos totales.

```python
pedidos = [23.50, 67.90, 45.00, 102.30, 12.99, 50.00, 8.75, 34.20, 89.99, 15.00, 60.50, 3.99, 77.00, 41.25, 55.55, 6.50, 92.10, 29.99, 48.00, 110.00]

ped_g = 0
pedidos_nog = 0

def envio_gratis(eur_ped):
    if eur_ped >= 50:
        return True
    else:
        return False

for p in pedidos:
    resultado = envio_gratis(p)
    if not resultado:
        pedidos_nog += 1
    else: 
        ped_g += 1
        
print(f'Envios gratuitos: {ped_g}')
print(f'Envios no gratuitos: {pedidos_nog}')
```

**Ejercicio 11: Lista de aprobados**

Un profesor tiene las notas de sus 25 alumnos (sobre 10). Se considera aprobado con nota 5 o más.

Escribe un programa que:

Cree una lista vacía llamada aprobados.
Recorra notas con un bucle for.
Cada vez que una nota sea >= 5, la añada a la lista aprobados.
Al final, muestre la lista aprobados completa y cuántos alumnos hay en ella.

```python
notas = [6.5, 3.2, 8.0, 4.5, 9.1, 5.0, 2.8, 7.3, 6.0, 4.9, 8.8, 3.5, 5.5, 9.9, 1.2, 6.8, 4.0, 7.0, 5.9, 3.0, 8.4, 6.2, 4.7, 9.0, 5.1]

alu_aprobados = 0
alu_suspendidos = 0

aprobados = []

for a in notas: 
    if a >= 5:
        aprobados.append(a)
        alu_aprobados += 1
    else:
        alu_suspendidos += 1
        
print(f'Alumnos suspendidos: {alu_suspendidos}')
print(f'Alumnos aprobados: {len(aprobados)}')
print(aprobados)
    
```

**Ejercicio 12: Informe de temperaturas**

Una estación meteorológica registró la temperatura (en ºC) cada hora durante un día completo (24 registros).

Escribe un programa que:

Calcule la temperatura media del día (suma total dividida entre el número de registros).
Encuentre la temperatura máxima y la mínima (sin usar max()/min()).
Cuente cuántas horas hubo con temperatura por encima de la media.
Muestre los 4 resultados con mensajes claros.

```python
temperaturas = [12.5, 11.8, 11.2, 10.9, 10.5, 11.0, 13.2, 15.8, 18.4, 21.0, 23.5, 25.8, 27.2, 28.1, 27.9, 26.5, 24.3, 21.7, 19.2, 17.5, 16.1, 15.0, 14.2, 13.5]

average_temp = sum(temperaturas) / len(temperaturas)
max = temperaturas[0]
min = temperaturas[0]
above_average = 0

for t in temperaturas:
    if t > max: max = t
    if t < min: min = t
    if t > average_temp:
        above_average += 1
    

print(f'La temperatura máxima fue {max}')
print(f'La temperatura minima fue {min}')
print(f'La temperatura media fue {average_temp}')
print(f'Hubo {above_average} horas por encima de la media')




        
```

**Ejercicio 15: Totales por vendedor**

Usando la misma lista ventas del ejercicio anterior, escribe un programa que calcule el total vendido por cada vendedor (no solo Ana), sin repetir código para cada nombre.
Dentro del bucle, por cada venta tienes el nombre y el importe. La pregunta clave es: ¿ese nombre ya existe como clave en el diccionario totales?

Si no existe todavía, tienes que crearlo con el importe de esta venta.
Si ya existe, tienes que sumarle el importe de esta venta al valor que ya tenía.

Para comprobar si una clave existe en un diccionario, puedes usar: if nombre in totales:

Al final, muestra el diccionario completo, o recórrelo con un for mostrando cada nombre y su total.

```python
ventas = [
    ["Ana", 1200], ["Luis", 890], ["Ana", 450], ["Marta", 2300],
    ["Luis", 670], ["Carlos", 1500], ["Ana", 980], ["Marta", 340],
    ["Carlos", 210], ["Luis", 1100], ["Ana", 760], ["Marta", 1890],
    ["Carlos", 950], ["Luis", 430], ["Marta", 600],
]

totales = {}

for n in ventas:
    nombre = n[0]
    importe = n[1]
    totales[nombre] = totales.get(nombre, 0) + importe

for nombre, importe in totales.items():
    print(f'{nombre} vendió {importe}€')
```
