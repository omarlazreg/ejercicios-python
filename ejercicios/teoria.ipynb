```python
txt = "We are the so-called \"Vikings\" from the north." # Illegal characters
print(txt)

# \' Single Quote
print('Está diciendo \'hola\'')

# \\ Backslash
print('La ruta es C:\\Users\\Pedro')
# Salida: La ruta es C:\Users\Pedro

# \n New Line
print('Primera línea\nSegunda línea')
# Salida:
# Primera línea
# Segunda línea

# \r Carriage Return
print('Hola Mundo\rAdiós')
# Salida: Adiósundo
# (el cursor vuelve al inicio y sobreescribe)

# \t Tab
print('Nombre:\tJuan')
# Salida: Nombre:    Juan

# \b Backspace
print('Hola\bo')
# Salida: Holo
# (\b borra el carácter anterior, la 'a' desaparece)

# \f Form Feed
print('Página1\fPágina2')
# Salida: Página1Página2
# (en consola normal casi no se nota, es un salto de página para impresoras/algunos terminales)

# \ooo Octal value
print('\110\145\154\154\157')
# Salida: Hello
# (cada código octal representa un carácter ASCII: 110=H, 145=e, etc.)

# \xhh Hex value
print('\x48\x65\x6c\x6c\x6f')
# Salida: Hello
# (mismo resultado que el octal, pero en hexadecimal)
```
