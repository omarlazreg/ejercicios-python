```python
import pandas as pd

mydataset = {
  'cars': ["BMW", "Volvo", "Ford"],
  'passings': [3, 7, 2]
}

myvar = pd.DataFrame(mydataset)

print(myvar)
```

```python
# Pandas series is like a column in a table

a = [4, 7, 2]
myvar = pd.Series(a)
print(myvar)

print(myvar[0]) # if nothing else is specified, the values are valued with their index number.
```

```python
# Create your own labels:

a = [1, 7, 2]
myvar = pd.Series(a, index = ["x", "y", "z"])
print(myvar)

print(myvar["y"])
```

```python
# Create a panda series from dictionary: 

calories = {"day1": 420, "day2": 380, "day3": 390}
myvar = pd.Series(calories)
print(myvar)
```

```python
calories = {"day1": 420, "day2": 380, "day3": 390}
myvar = pd.Series(calories, index = ["day1", "day3"])
print(myvar)
```

```python
# Create a dataframe

data = {
  "calories": [420, 380, 390],
  "duration": [50, 40, 45]
}

#load data into a DataFrame object:
df = pd.DataFrame(data)
print(df);print()

#refer to the row index:
print(df.loc[0]); print()
print(df.loc[[0, 1]])
```

```python
# Add a list of names to give each row a name:

data = {
  "calories": [420, 380, 390,240],
  "duration": [50, 40, 45,50]
}

df = pd.DataFrame(data, index = ["day1", "day2", "day3", 'day4'])

print(df); print()
print(df.loc["day2"])
```
