```python
import datetime
import json


datos = [
    'Título de tarea: ',
    'Fecha: ',
    'Hora: ',
    'Lugar/ubicación: ',
    'Observaciones:',
    'Asistentes: ',
    '¿Periódico?:', 
    'Urgencia: '
]

all_tasks = []

while True:
    tareas = {}
    for t in datos: 
        tareas[t] = input(f'{t} ')
    all_tasks.append(tareas)
    respuesta = input('¿Quieres añadir otra tarea? (s/n): ')
    if respuesta == 'n':
        break
with open('tasks.json', 'w') as f:
    json.dump(all_tasks, f)
print(all_tasks)
```
